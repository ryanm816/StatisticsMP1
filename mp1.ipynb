```python
rng = np.random.default_rng(5)

def bsc(bits, p, rng):
    #my idea: randomly generate which positions are flipped.  Array of normal bits and rern array

    flippedBitsNum = int(round(len(bits)*p,1))

    for i in range(flippedBitsNum):
        #generate random number
        flipBit = rng.integers(0, len(bits) + 1)
        #flip bit in array
        bits[flipBit] ^= 1  #using XOR to flip it!

    return bits

# this function above will be called from seperate doc.  Let's pass in array of just ones

# the function below will act like our circuit.  For this example. I'll consider 0s as errors


def error_circuit(bits):
    flag = 0
    flagInError = 0
    for i in range(len(bits)):
        if bits[i] == 0:
            if rng.random() < 0.95:
                flag += 1
                flagInError += 1  #actually in error
        else:
           if rng.random() < 0.02:
                flag += 1
    #return ratio of flagged & in error to just flagged 
    ratio = flagInError/flag
    return ratio

array = np.ones(10**7, dtype=int)
ratio = error_circuit(bsc(array, 0.0206, rng))

            
print("The ratio of flagged bits in error to flagged bits is", round(ratio, 4)) 
```

If there is a good link and the number of bits in error is very low, the odds of the flag actually signifying an error is also very low.
```python
import numpy as np
import pandas as pd
from scipy.stats import binom, poisson
#Computing exact Binomial pmf and Poisson pmf with same mean and their distance under 5 cases to print a table 
n = 1000
p_values = [0.5, 0.05, 0.005, 0.001, 0.0005]
results = []
for p in p_values:
#Equation for truncating the PMF
    #Poisson pmf mean
    lam = n * p
    K = int(np.ceil(lam + 20*np.sqrt(lam) + 20))
    k = np.arange(K + 1)
#Computing exact Binomial pmf and Poisson pmf with same mean
    binom_pmf = binom.pmf(k, n, p)
    poisson_pmf = poisson.pmf(k, lam)
#compute distance for the five cases
    distance = tv_distance(binom_pmf, poisson_pmf)
    """compute remaining poisson probability tail beyond K to print next to each ditance and show 
    that the truncation didn't matter"""
    poisson_tail = poisson.sf(K, lam)
#place the 5 cases in a list called results
    results.append((p, lam, K, distance, poisson_tail))
# Print a table for the data
distance_table = pd.DataFrame(results, columns=["p", "lambda = np", "K", "TV Distance", "Poisson Tail > K"])
print(distance_table.to_string(index=False))
```

```python
#Finding the value of p for which for which the distance stays below .01 using bisection 
def tv_for_p(p, n=1000):
    lam = n * p
    K = int(np.ceil(lam + 20*np.sqrt(lam) + 20))
    k = np.arange(K + 1)
    binom_pmf = binom.pmf(k, n, p)
    poisson_pmf = poisson.pmf(k, lam)
    return tv_distance(binom_pmf, poisson_pmf)
#setting the bracket of p from .001 to .1 to find the value of p for which the distance stays below .01
p_low = 0.001
p_high = 0.1
for _ in range(50):
    p_mid = (p_low + p_high) / 2
    d_mid = tv_for_p(p_mid)
    if d_mid < 0.01:
        p_low = p_mid
    else:
        p_high = p_mid
p_threshold = p_low
tv_threshold = tv_for_p(p_threshold)
#Print the value of the largest p with TV distance less than .01 to two sig figs
print("Largest p with TV distance < 0.01:")
print(f"p = {p_threshold:.3f}")
print(f"TV distance = {tv_threshold:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
SEED = 5
rng = np.random.default_rng(SEED)
rng_step7 = rng.spawn(1)[0]

# Parameter
N_bits = 1000
p_error = 0.005
q = 0.7
num_runs = 200
# 30 log-spaced checkpoints from 10 to 10^5
checkpoints = np.unique(np.logspace(1, 5, 30).astype(int))

# Binomial from step 4
binomial_mean_theory = N_bits * p_error
binomial_var_theory = N_bits * p_error * (1 - p_error)

# Geometric from step 6
geometric_mean_theory = 1 / q
geometric_var_theory = (1 - q) / (q**2)

print("Binomial Distribution")
print("Theoretical Mean =", round(binomial_mean_theory, 6))
print("Theoretical Variance =", round(binomial_var_theory, 6))
print()
print("Geometric Distribution")
print("Theoretical Mean =", round(geometric_mean_theory, 6))
print("Theoretical Variance =", round(geometric_var_theory, 6))
print()
print("Number of checkpoints =", len(checkpoints))
print("Checkpoints:")
print(checkpoints)
```

```python
# Run 200 Independent Experiments
binomial_running_means = np.zeros((num_runs, len(checkpoints)))

geometric_running_means = np.zeros((num_runs, len(checkpoints)))
for run in range(num_runs):
    binomial_total = 0.0
    geometric_total = 0.0
    previous_N = 0
    for j, N in enumerate(checkpoints):     
        block_size = N - previous_N # Number of new samples required
        binomial_samples = rng_step7.binomial(
            n=N_bits,
            p=p_error,
            size=block_size
        )
        binomial_total += np.sum(binomial_samples)
        binomial_running_means[run, j] = (
            binomial_total / N
        )
        geometric_samples = rng_step7.geometric(
            p=q,
            size=block_size
        )
        geometric_total += np.sum(geometric_samples)
        geometric_running_means[run, j] = (
            geometric_total / N
        )
        previous_N = N

print("Stored Binomial array shape:", binomial_running_means.shape)
print("Stored Geometric array shape:", geometric_running_means.shape)
```

```python
# Average running mean 
binomial_average = np.mean(binomial_running_means, axis=0)
geometric_average = np.mean(geometric_running_means, axis=0)

# Use one standard deviation
binomial_spread = np.std(binomial_running_means, axis=0, ddof=1)
geometric_spread = np.std(geometric_running_means, axis=0, ddof=1)

# Find the checkpoint closest to N = 1000
match_index = np.argmin(np.abs(checkpoints - 1000))
N_match = checkpoints[match_index]

# spread at N approximately equal to 1000
binomial_scale = (binomial_spread[match_index] * np.sqrt(N_match))
geometric_scale = (geometric_spread[match_index] * np.sqrt(N_match))

binomial_envelope = (binomial_scale / np.sqrt(checkpoints))
geometric_envelope = (geometric_scale / np.sqrt(checkpoints))

print("1/sqrt(N) envelope matched at N =", N_match)
```

At smaller values of N, the shaded regions are relatively wide, indicating greater variability in the sample mean estimates. As N increases, the running sample means approach their theoretical values, and the shaded regions become narrower.

Bonus A

From Step 6, the number of attempts required to receive an intact packet is a random variable following a geometric distribution. Let T be the number of attempts (including the successful one), and let q = 0.7 be the probability of receiving an intact packet in each attempt. The attempts are assumed to be independent of one another.

$$
P(T=t)=q(1-q)^{t-1},\qquad t=1,2,3,\ldots
$$

The event T>k occurs only if the first k attempts all fail.

$$
\{T>k\}=F_1\cap F_2\cap\cdots\cap F_k.
$$

Using independence across attempts,

$$
P(T>k)=P(F_1)P(F_2)\cdots P(F_k)
={(1-q)(1-q)\cdots(1-q)}
$$

$$
{P(T>k)=(1-q)^k=(0.3)^k.}
$$

Because an intersection of failures is contained in their union,

$$
F_1\cap\cdots\cap F_k \subseteq F_1\cup\cdots\cup F_k.
$$

The union bound states

$$
P(F_1\cup\cdots\cup F_k)\le\sum_{i=1}^k P(F_i)
=k(1-q).
$$

Therefore, a valid but loose upper bound on the event T>k is

$$
\boxed{P(T>k)\le\min\left(1,\,k(1-q)\right)=\min(1,0.3k).}
$$

```python
bonusA_q = 0.7
bonusA_N = 10**6
bonusA_k = np.arange(1, 8)

bonusA_attempts = attempts_until_clean(bonusA_N, bonusA_q, rng_bonusA)
bonusA_exact = (1 - bonusA_q) ** bonusA_k
bonusA_sim = np.array([np.mean(bonusA_attempts > k) for k in bonusA_k])
bonusA_difference = np.abs(bonusA_sim - bonusA_exact)
bonusA_se = np.sqrt(bonusA_sim * (1 - bonusA_sim) / bonusA_N)

# At least one failure is the union event; all failures is the tail event.
bonusA_any_failure = 1 - bonusA_q ** bonusA_k
bonusA_bound = np.minimum(1, bonusA_k * (1 - bonusA_q))

print('BONUS A - GEOMETRIC TAIL AND UNION BOUND')
print('q =', bonusA_q)
print('Number of simulated packets =', bonusA_N)
print()
print(f"{'k':<4}{'Exact P(T>k)':<16}{'Simulation':<16}{'Difference':<16}{'Std. Error':<16}{'Union Bound':<16}")
print('_' * 84)
for i, k in enumerate(bonusA_k):
    print(f'{k:<4}{bonusA_exact[i]:<16.6f}{bonusA_sim[i]:<16.6f}'
          f'{bonusA_difference[i]:<16.6f}{bonusA_se[i]:<16.6f}{bonusA_bound[i]:<16.6f}')

print('\nFor k=3: exact P(T>3) =', f'{bonusA_exact[2]:.6f}', '; union bound =', f'{bonusA_bound[2]:.6f}')
```

```python
fig, ax = plt.subplots(figsize=(9, 6))
ax.semilogy(bonusA_k, bonusA_exact, 'o-', label='Exact P(T > k)')
ax.errorbar(bonusA_k, bonusA_sim, yerr=bonusA_se,
            fmt='s', capsize=3, label='Simulation Estimate')
ax.semilogy(bonusA_k, bonusA_any_failure, 'd-.',
            label='P(at least one failure)')
ax.semilogy(bonusA_k, bonusA_bound, '^--',
            label='Union bound')
ax.set_xlabel('Number of attempts k')
ax.set_ylabel('Probability (log scale)')
ax.set_title('Bonus A: Geometric Tail and Union Bound')
ax.set_xticks(bonusA_k)
ax.grid(True, which='both', alpha=0.25)
ax.legend()
plt.show()
```

The graph compares the theoretical probabilities, simulated probabilities, probability of at least one failure, and the union bound for the geometric distribution with a successful transmission probability of $q=0.7$.

The blue solid line represents the theoretical probability that a packet requires more than $k$ attempts:

$$
P(T>k)=(1-q)^k=(0.3)^k.
$$

As the number of attempts increases, the probability decreases exponentially because all of the first $k$ transmission attempts must fail.

The orange squares represent the simulated probabilities with their standard errors. The simulated results closely follow the theoretical curve, showing that the experimental results are consistent with the geometric probability model.

The green dashed line represents the probability that at least one of the first $k$ transmission attempts fails:

$$
P(F_1\cup\cdots\cup F_k)=1-q^k.
$$

This probability increases as $k$ increases because more transmission attempts provide more opportunities for at least one failure to occur.

The red dashed line represents the union bound:

$$
P(T>k)\leq\min(1,k(1-q)).
$$

The union bound becomes increasingly loose as $k$ increases. For example, when $k=3$, the theoretical probability of requiring more than three attempts is

$$
P(T>3)=(0.3)^3=0.027,
$$

while the union bound is

$$
3(0.3)=0.9.
$$

The difference occurs because the exact geometric probability requires all of the first k attempts to fail, while the union-bound comparison considers the less restrictive event that at least one attempt fails. Additionally, the union bound can overestimate the probability by counting overlapping failure events multiple times.


Bonus B

From Step 4, each bit is independently corrupted with probability $p$.

The probability that one bit is transmitted correctly is

$$
P(\text{Correct Bit})=1-p.
$$

Since all bits are independent, the probability that an entire packet
of length $L$ arrives without errors is

$$
q(L)=(1-p)^L.
$$

For $p=10^{-4}$,

$$
{q(L)=(1-10^{-4})^L}
$$

From Step 6, the number of attempts required until the first clean
packet follows a geometric distribution.

$$
T\sim\text{Geometric}(q)
$$

The theoretical expected value is

$$
E[T]=\frac{1}{q}.
$$

Substituting the clean-packet probability,

$$
E[T]=\frac{1}{(1-p)^L}.
$$

```python
p = 1e-4
header = 40

# Search integer packet lengths greater than the header
L_values = np.arange(header + 1, 5001)

# Probability that the packet is clean
q_values = (1 - p) ** L_values

# Goodput equation from Steps 4, 6, and 8
G_values = ((L_values - header) / L_values) * q_values

best_index = np.argmax(G_values)

L_best = int(L_values[best_index])
G_best = G_values[best_index]

print("NUMERICAL OPTIMIZATION")
print()

print(f"Bit Error Probability = {p:.6f}")
print(f"Header Length = {header} bits")
print(f"Optimal Packet Length = {L_best} bits")
print(f"Maximum Goodput = {G_best:.6f}")
print()
```

Goodput is defined as useful delivered bits divided by transmitted bits.

The number of useful payload bits is

$$
L-40.
$$

The expected number of transmitted bits required to deliver one packet is

$$
L E[T]=\frac{L}{q(L)}.
$$

Therefore,

$$
G(L)=\frac{L-40}{L/q(L)}.
$$

$$
G(L)=\frac{L-40}{L}q(L).
$$

Substituting $q(L)=(1-p)^L$,

$$
{
G(L)=\left(1-\frac{40}{L}\right)(1-p)^L
}
$$

where $L>40$.

```python
a = -np.log1p(-p)

# Positive root of:
L_derivative = (header + np.sqrt(header**2 + 4 * header / a)) / 2

# Goodput at continuous optimum
G_derivative = ((L_derivative - header) / L_derivative * (1 - p) ** L_derivative)

# Verify derivative of ln(G) is approximately zero
derivative_check = (header / (L_derivative * (L_derivative - header)) + np.log1p(-p))

print("DERIVATIVE CHECK")
print()
print(f"Continuous Optimal Length = {L_derivative:.6f}")
print(f"Goodput at Continuous Optimum = {G_derivative:.6f}")
print(f"Derivative at Optimum = {derivative_check:.12f}")
print(f"Numerical Integer Optimum = {L_best}")
print(f"Length Difference = {abs(L_best - L_derivative):.6f}")
print()
```

To maximize goodput, differentiate the natural logarithm of $G(L)$.

$$
\ln G(L)=\ln(L-40)-\ln L+L\ln(1-p).
$$

Taking the derivative,

$$
\frac{d}{dL}\ln G(L)
=
\frac{1}{L-40}-\frac{1}{L}+\ln(1-p).
$$


$$
\frac{d}{dL}\ln G(L)
=
\frac{40}{L(L-40)}+\ln(1-p).
$$

At the maximum, set the derivative equal to zero.

$$
\frac{40}{L(L-40)}+\ln(1-p)=0.
$$



$$
\frac{40}{L(L-40)}=-\ln(1-p).
$$

Let

$$
a=-\ln(1-p).
$$



$$
L(L-40)=\frac{40}{a}.
$$



$$
L^2-40L-\frac{40}{a}=0.
$$

Using the quadratic formula,

$$
L=\frac{40+\sqrt{40^2+\frac{160}{a}}}{2}.
$$

Substituting $p=10^{-4}$ gives

$$
{L^*\approx652.756\text{ bits}}.
$$

Since the packet length must be an integer, the optimal packet length is

$$
\boxed{L=653\text{ bits}}.
$$

```python
# Selected packet lengths around and beyond the optimum
L_sim = np.array([
    200, 400, 500, 550, 600,
    650, 653, 700, 750, 800,
    900, 1200, 2000
])

# Number of independent delivered packets per length
N_sim = 200000

sim_goodput = []
sim_se = []

for L in L_sim:

    # Clean-packet success probability
    q_L = (1 - p) ** L

    # Geometric distribution:
    attempts = rng_bonusB.geometric(p=q_L, size=N_sim)

    # Sample mean of transmission attempts
    mean_attempts = np.mean(attempts)

    # Standard error of mean attempts
    se_attempts = (np.std(attempts, ddof=1) / np.sqrt(N_sim))

    # Fraction of each packet that is useful data
    payload_fraction = (L - header) / L

    # Simulation goodput:
    simulated_G = payload_fraction / mean_attempts

    # Approximate standard error of goodput
    standard_error = (payload_fraction * se_attempts / mean_attempts**2)

    sim_goodput.append(simulated_G)
    sim_se.append(standard_error)

sim_goodput = np.array(sim_goodput)
sim_se = np.array(sim_se)
```

```python
print("THEORY VS SIMULATION")
print()

print(
    f"{'L':<8}"
    f"{'Theory':<14}"
    f"{'Simulation':<14}"
    f"{'Difference':<14}"
    f"{'Std. Error':<14}"
)

print("-" * 64)

for i, L in enumerate(L_sim):

    theoretical_G = (
        (L - header) / L
        * (1 - p) ** L
    )

    difference = abs(
        sim_goodput[i] - theoretical_G
    )

    print(
        f"{L:<8}"
        f"{theoretical_G:<14.6f}"
        f"{sim_goodput[i]:<14.6f}"
        f"{difference:<14.6f}"
        f"{sim_se[i]:<14.6f}"
    )

print()
```
