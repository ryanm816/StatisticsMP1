Argue in two sentences why the number of errors in a 1000-bit packet is Binomial,
naming the assumption on the bits that makes it so.

We assume that all bits are independent and all have the same chance of being corrupted. It is binomial because the only options are pass and fail and one bit failing would not effect the bits that come after it which means it follows binomial distribution. 

```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.stats import binom
from mp1lib import packet_errors 
```

```python
# unique group seed 
SEED = 5 
rng = np.random.default_rng(SEED)
rng4 = rng.spawn(1)[0]

#pararmeters
N_pkt = 100000
N_bits = 1000
P_error = .005

errors = packet_errors( 
    n_pkt = N_pkt,
    n_bits = N_bits,
    p = P_error,
    rng  = rng4
)

#calculations 
sample_mean = np.mean(errors)
sample_var = np.var(errors, ddof =1)
theoretical_mean = N_bits * P_error
theoretical_var = N_bits * P_error * (1 - P_error) 
mean_diff = sample_mean - theoretical_mean
var_diff = sample_var - theoretical_var
mean_error = np.sqrt(theoretical_var / N_pkt)

print("Bernouli trials and the binomial") 
print(f"Sample mean:      {sample_mean:.6f}")
print(f"Theoretical mean: {theoretical_mean:.6f}")
print(f"Mean Difference:  {mean_diff:.6f}")
print()
print(f"Sample Variance: {sample_var:.6f}")
print(f"Theoretical Variance: {theoretical_var:.6f}")
print(f"Variance: {var_diff:.6f}")

#plot
plt.figure(figsize = (9,6)) 
plt.hist(
    errors,
    bins = np.arange(-.5, errors.max() +2, 1),
    density = True,
    alpha = .75,
    label = "Simulation"
)

k_values = np.arange(0,errors.max()+1) 

binomial_pmf = binom.pmf(
    k_values,
    n = N_bits,
    p = P_error
)

plt.plot(
    k_values,
    binomial_pmf,
    markersize = 5,
    label = "Binomial PMF"
)

plt.xlabel("Errors")
plt.ylabel("Probability")
plt.title("Figure 3 Binomial Error Distribution")
plt.legend()
plt.grid(alpha = .05)
plt.show()

print()
print("Clean Packet Probabilities")

p_val = [.005, .001, .0001]
for p in p_val: 
    packet_error_clean = packet_errors(
    n_pkt = N_pkt,
    n_bits = N_bits,
    p = p,
    rng = rng4
    )
    theoretical_clean = (1 - p) ** N_bits
    emperical_clean = np.mean(packet_error_clean == 0)
    difference = emperical_clean - theoretical_clean
    Error = np.sqrt( theoretical_clean * (1 -theoretical_clean) / N_pkt)
    
    print(f"p = {p}")
    print(f"Theoretical Probability: {theoretical_clean:.6f}")
    print(f"Emperical Probability:   {emperical_clean:.6f}")
    print(f"Difference:              {difference:.6f}")
    print(f"Error:                   {Error:.6f}")
    print()
```

The graph shows distribution of error when transmitting 100000 packets of 1000 bits. Each blue bar represents the probability of that many errors being present in a packet while the orange curve shows the mathemetical value of the theoretical PMF.
