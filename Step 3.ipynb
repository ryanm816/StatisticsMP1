### Step 3 - Part 1: Pairwise Independence and Mutual Independence

For two fair coin flips, the sample space is

$$
S = \{HH, HT, TH, TT\}
$$

Since both coins are fair, each outcome has probability

$$
P(HH)=P(HT)=P(TH)=P(TT)=\frac{1}{4}
$$

### Define the three events.
E1 is the event that the first coin is Heads.

$$
E_1 = \{HH, HT\}
$$

$$
P(E_1)=\frac{2}{4}=\frac{1}{2}
$$


E2 is the event that the second coin is Heads.

$$
E_2 = \{HH, TH\}
$$

$$
P(E_2)=\frac{2}{4}=\frac{1}{2}
$$

E3 is the event that the two coins match.

$$
E_3 = \{HH, TT\}
$$

$$
P(E_3)=\frac{2}{4}=\frac{1}{2}
$$

---

### Check independence of E1 and E2

The intersection is

$$
E_1 \cap E_2 = \{HH\}
$$

$$
P(E_1 \cap E_2)=\frac{1}{4}
$$

The product of the individual probabilities is

$$
P(E_1)P(E_2)
=
\left(\frac{1}{2}\right)
\left(\frac{1}{2}\right)
=
\frac{1}{4}
$$


$$
P(E_1 \cap E_2)=P(E_1)P(E_2)
$$

\(E1\) and \(E2\) are independent.

---

### Check independence of E1 and E3

The intersection is

$$
E_1 \cap E_3 = \{HH\}
$$


$$
P(E_1 \cap E_3)=\frac{1}{4}
$$

The product of the individual probabilities is

$$
P(E_1)P(E_3)
=
\left(\frac{1}{2}\right)
\left(\frac{1}{2}\right)
=
\frac{1}{4}
$$

$$
P(E_1 \cap E_3)=P(E_1)P(E_3)
$$

\(E1\) and \(E3\) are independent.

---

### Check independence of E2 and E3

The intersection is

$$
E_2 \cap E_3 = \{HH\}
$$


$$
P(E_2 \cap E_3)=\frac{1}{4}
$$

The product of the individual probabilities is

$$
P(E_2)P(E_3)
=
\left(\frac{1}{2}\right)
\left(\frac{1}{2}\right)
=
\frac{1}{4}
$$


$$
P(E_2 \cap E_3)=P(E_2)P(E_3)
$$

\(E2\) and \(E3\) are independent.

Therefore, all three pairs are independent, so the events are pairwise independent.

---

## Check Mutual Independence

For three events to be mutually independent, they must satisfy

$$
P(E_1 \cap E_2 \cap E_3)
=
P(E_1)P(E_2)P(E_3)
$$

The triple intersection is

$$
E_1 \cap E_2 \cap E_3 = \{HH\}
$$

$$
P(E_1 \cap E_2 \cap E_3)=\frac{1}{4}
$$


$$
P(E_1)P(E_2)P(E_3)
=
\left(\frac{1}{2}\right)
\left(\frac{1}{2}\right)
\left(\frac{1}{2}\right)
=
\frac{1}{8}
$$


$$
\frac{1}{4} \neq \frac{1}{8}
$$


$$
P(E_1 \cap E_2 \cap E_3)
\neq
P(E_1)P(E_2)P(E_3)
$$

Therefore, the three events are not mutually independent.

---

### Conclusion

The events \(E_1\), \(E_2\), and \(E_3\) are pairwise independent because every pair satisfies

$$
P(A \cap B)=P(A)P(B)
$$

However, they are not mutually independent because

$$
P(E_1 \cap E_2 \cap E_3)
\neq
P(E_1)P(E_2)P(E_3)
$$

Therefore, pairwise independence does not imply mutual independence.

```python
import numpy as np

seed = 5       
rng = np.random.default_rng(seed)

rng_step3 = rng.spawn(1)[0]
N = 10**6

# 1 = Heads, 0 = Tails
coin1 = rng_step3.integers(0, 2, size=N)
coin2 = rng_step3.integers(0, 2, size=N)

E1 = (coin1 == 1)          # First coin is heads
E2 = (coin2 == 1)          # Second coin is heads
E3 = (coin1 == coin2)      # Two coins match

P_E1 = np.mean(E1)
P_E2 = np.mean(E2)
P_E3 = np.mean(E3)

P_E1E2 = np.mean(E1 & E2)
P_E1E3 = np.mean(E1 & E3)
P_E2E3 = np.mean(E2 & E3)
P_E1E2E3 = np.mean(E1 & E2 & E3)

# Products
prod_E1E2 = P_E1 * P_E2
prod_E1E3 = P_E1 * P_E3
prod_E2E3 = P_E2 * P_E3
prod_E1E2E3 = P_E1 * P_E2 * P_E3

print("STEP 3 - PART 1")
print()

print("Individual probabilities:")
print("P(E1) =", P_E1)
print("P(E2) =", P_E2)
print("P(E3) =", P_E3)
```

```python
print(f"{'Comparison':<20} {'Joint':<15} {'Product':<15} {'Difference':<15}")
print("-" * 65)

print(f"{'E1 and E2':<20} "
      f"{P_E1E2:<15.6f} "
      f"{prod_E1E2:<15.6f} "
      f"{abs(P_E1E2-prod_E1E2):<15.6f}")

print(f"{'E1 and E3':<20} "
      f"{P_E1E3:<15.6f} "
      f"{prod_E1E3:<15.6f} "
      f"{abs(P_E1E3-prod_E1E3):<15.6f}")

print(f"{'E2 and E3':<20} "
      f"{P_E2E3:<15.6f} "
      f"{prod_E2E3:<15.6f} "
      f"{abs(P_E2E3-prod_E2E3):<15.6f}")

print(f"{'E1,E2,E3':<20} "
      f"{P_E1E2E3:<15.6f} "
      f"{prod_E1E2E3:<15.6f} "
      f"{abs(P_E1E2E3-prod_E1E2E3):<15.6f}")
```

```python
N = 10**6

X = rng_step3.integers(0, 2, size=N)

error1 = rng_step3.random(N) < 0.1
error2 = rng_step3.random(N) < 0.3

Y1 = X ^ error1.astype(int)
Y2 = X ^ error2.astype(int)
r_all = np.corrcoef(Y1, Y2)[0, 1]

mask0 = (X == 0)
r_X0 = np.corrcoef(Y1[mask0], Y2[mask0])[0, 1]

mask1 = (X == 1)
r_X1 = np.corrcoef(Y1[mask1], Y2[mask1])[0, 1]

def correlation_se(r, n):
    return (1 - r**2) / np.sqrt(n - 3)

se_all = correlation_se(r_all, N)
se_X0 = correlation_se(r_X0, np.sum(mask0))
se_X1 = correlation_se(r_X1, np.sum(mask1))

print("STEP 3 - PART 2")
print()

print(f"{'Condition':<22} {'Samples':<12} {'Correlation':<15} {'Std. Error':<15}")
print("-" * 63)

print(f"{'All transmissions':<22} "
      f"{N:<12} "
      f"{r_all:<15.6f} "
      f"{se_all:<15.6f}")

print(f"{'Given X = 0':<22} "
      f"{np.sum(mask0):<12} "
      f"{r_X0:<15.6f} "
      f"{se_X0:<15.6f}")

print(f"{'Given X = 1':<22} "
      f"{np.sum(mask1):<12} "
      f"{r_X1:<15.6f} "
      f"{se_X1:<15.6f}")
```
